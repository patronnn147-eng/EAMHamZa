# Demo: Load and Run the ML Model

This notebook shows how to load the trained `basic_machine_model.pkl` and make a prediction using a sample telemetry vector.


```python
import os
import joblib
import numpy as np

model_path = os.path.join(os.getcwd(),'app','backend','modules','ml','basic_machine_model.pkl')
model = joblib.load(r"C:\Users\Admin\Downloads\EAM\EAMSagemCom\app\backend\modules\ml\basic_machine_model.pkl")
print('Model loaded')
```

```python
# Example feature vector (order: Air temp, Process temp, Rotational speed, Torque, Tool wear)
sample = np.array([[300.0, 310.0, 1500.0, 40.0, 5.0]])

pred = model.predict(sample)[0]
prob = model.predict_proba(sample)[0,1] * 100
print(f'Prediction: {"FAILURE" if pred==1 else "HEALTHY"}')
print(f'Failure probability: {prob:.1f}%')
```
