# 🛠️ EAM Predictive Maintenance Dashboard

Welcome to the consolidated testing environment for the ML Pipeline. This notebook allows you to:
1. **P1 (Failure Prediction)**: Predict the overall probability of a machine failure.
2. **P2 (Failure Type)**: Classify the specific root cause (TWF, HDF, PWF, OSF, RNF).
3. **P3 (RUL Estimation)**: Estimate the Remaining Useful Life (RUL) in cycles.
4. **P5 (Work Order Priority)**: Predict the urgency of the maintenance action.

---

```python
import os
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report, mean_absolute_error

# Set style for premium look
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams['figure.figsize'] = (10, 6)

print("✅ Setup Complete.")
```

## 📂 Load Data and Models

```python
# --- Paths ---
BASE_PATH = "."
MODELS_DIR = "app/backend/modules/ml/models"
CSV_PATH = "ai4i2020.csv"

# --- Load Dataset ---
df = pd.read_csv(CSV_PATH)
print(f"📊 Dataset loaded: {df.shape[0]} rows")

# --- Load Models ---
try:
    model_p1 = joblib.load(os.path.join(MODELS_DIR, "basic_machine_model.pkl"))
    print("✅ P1 Model Loaded (Failure Prediction)")
    
    data_p2 = joblib.load(os.path.join(MODELS_DIR, "ml_model_p2_failure_type.pkl"))
    model_p2 = data_p2['model']
    p2_labels = data_p2['labels']
    print("✅ P2 Model Loaded (Failure Type)")
    
    model_p3 = joblib.load(os.path.join(MODELS_DIR, "ml_model_p3_rul.pkl"))
    print("✅ P3 Model Loaded (RUL Estimation)")

    data_p5 = joblib.load(os.path.join(MODELS_DIR, "ml_model_p5_priority.pkl"))
    model_p5 = data_p5['model']
    p5_labels = data_p5['labels']
    print("✅ P5 Model Loaded (Work Order Priority)")
except Exception as e:
    print(f"❌ Error loading models: {e}")
```

## 🖥️ Prediction Dashboard

Use the cell below to test any telemetry snapshot. It will run through all three models simultaneously.

```python
def run_ml_diagnostics(air_temp, process_temp, rpm, torque, wear):
    # 1. Prepare Inputs
    # P1/P3/P5 use: [air, process, rpm, torque, wear]
    # P2 uses: [air, process, rpm, torque, wear, temp_delta]
    temp_delta = process_temp - air_temp
    input_p135 = np.array([[air_temp, process_temp, rpm, torque, wear]])
    input_p2 = np.array([[air_temp, process_temp, rpm, torque, wear, temp_delta]])

    # 2. Run Predictions
    prob_fail = model_p1.predict_proba(input_p135)[0, 1] * 100
    predicted_rul = model_p3.predict(input_p135)[0]
    p2_preds = model_p2.predict(input_p2)[0]
    p2_probs = [est.predict_proba(input_p2)[0, 1] * 100 for est in model_p2.estimators_]
    
    # P5 Prediction (Priority)
    priority_idx = model_p5.predict(input_p135)[0]
    priority_label = p5_labels[priority_idx]

    # 3. Print Professional Report
    print("="*50)
    print("🚀 MACHINE HEALTH DIAGNOSTIC REPORT")
    print("="*50)
    print(f"INPUTS: Temp Ratio: {temp_delta:.1f}K | RPM: {rpm} | Torque: {torque}Nm | Wear: {wear}min")
    print("-"*50)
    
    # P1 Result
    status_p1 = "⚠️  DANGER" if prob_fail > 50 else "✅ HEALTHY"
    print(f"[P1] FAILURE RISK         : {prob_fail:>6.1f}% -> {status_p1}")
    
    # P3 Result
    print(f"[P3] EST. REMAINING LIFE  : {predicted_rul:>6.1f} cycles")
    
    # P5 Result
    print(f"[P5] WO PRIORITY         :    {priority_label.upper()}")
    
    # P2 Result
    print("-"*50)
    print("[P2] ROOT CAUSE ANALYSIS:")
    for label, detected, p2_prob in zip(p2_labels, p2_preds, p2_probs):
        mark = "[X]" if detected == 1 else "[ ]"
        print(f"    {mark} {label:3s}: {p2_prob:>5.1f}% confidence")
    print("="*50)

# --- TEST CASE: Stressed Machine ---
run_ml_diagnostics(air_temp=302.5, process_temp=312.0, rpm=1350, torque=62.0, wear=235)

# --- TEST CASE: Healthy Machine ---
run_ml_diagnostics(air_temp=298.0, process_temp=308.0, rpm=1550, torque=38.0, wear=15)
```

## 📊 Model Diagnostics & Analytics

This section visualizes exactly how the models are thinking.

```python
# 1. Feature Importance (P1)
importances = model_p1.feature_importances_
features = ['Air Temp', 'Process Temp', 'RPM', 'Torque', 'Tool Wear']

plt.figure(figsize=(10, 5))
sns.barplot(x=importances, y=features, hue=features, palette='magma', legend=False)
plt.title("Which sensors drive Failure Predictions?")
plt.xlabel("Importance Score")
plt.show()

# 2. Failure Distribution (Data Insight)
plt.figure(figsize=(8, 4))
sns.countplot(x='Machine failure', data=df, hue='Machine failure', palette='Set2', legend=False)
plt.title("Dataset Balance (Healthy vs Failed)")
plt.show()
```

```python
# 3. RUL Correlation Heatmap
# Let's see how RUL correlates with sensor data
corr_cols = ['Air temperature [K]', 'Process temperature [K]', 'Rotational speed [rpm]', 'Torque [Nm]', 'Tool wear [min]', 'Machine failure']
plt.figure(figsize=(10, 8))
sns.heatmap(df[corr_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Sensor Correlations")
plt.show()
```
